```r
df_scenario1 <- read.csv("scenario1_simulations.csv")
df_scenario2 <- read.csv("scenario2_simulations.csv")
df_scenario3 <- read.csv("scenario3_simulations.csv")

df_scenario1$true_value <- rep(c(0.1,0.1,0.8), 500)
df_scenario1$bias <- df_scenario1$mean - df_scenario1$true_value
df_scenario1$mse <- (df_scenario1$mean - df_scenario1$true_value)**2

df_scenario2$true_value <- rep(c(0.1,0.7,0.2), 500)
df_scenario2$bias <- df_scenario2$mean - df_scenario2$true_value
df_scenario2$mse <- (df_scenario2$mean - df_scenario2$true_value)**2

df_scenario3$true_value <- rep(c(0.1,0.3,0.3), 500)
df_scenario3$bias <- df_scenario3$mean - df_scenario3$true_value
df_scenario3$mse <- (df_scenario3$mean - df_scenario3$true_value)**2

head(df_scenario1)
head(df_scenario2)
head(df_scenario3)
```

### Scenario 1: $\alpha_0 = 0.1$, $\alpha_1 = 0.1$, $\beta_1 = 0.8$

```r
# Scenario 1
true_alpha0 <- 0.1
true_alpha1 <- 0.1
true_beta1 <- 0.8

alpha0_lower_500 <- df_scenario1[ (df_scenario1$parameter == "alpha0") & (df_scenario1$T == 500), 7]
alpha0_upper_500 <- df_scenario1[ (df_scenario1$parameter == "alpha0") & (df_scenario1$T == 500), 11]
alpha0_lower_750 <- df_scenario1[ (df_scenario1$parameter == "alpha0") & (df_scenario1$T == 750), 7]
alpha0_upper_750 <- df_scenario1[ (df_scenario1$parameter == "alpha0") & (df_scenario1$T == 750), 11]
alpha0_lower_1000 <- df_scenario1[ (df_scenario1$parameter == "alpha0") & (df_scenario1$T == 1000), 7]
alpha0_upper_1000 <- df_scenario1[ (df_scenario1$parameter == "alpha0") & (df_scenario1$T == 1000), 11]
coverage_prob_alpha0_500 <- mean( (alpha0_lower_500 < true_alpha0) & (alpha0_upper_500 > true_alpha0) )
coverage_prob_alpha0_750 <- mean( (alpha0_lower_750 < true_alpha0) & (alpha0_upper_750 > true_alpha0) )
coverage_prob_alpha0_1000 <- mean( (alpha0_lower_1000 < true_alpha0) & (alpha0_upper_1000 > true_alpha0) )

alpha1_lower_500 <- df_scenario1[ (df_scenario1$parameter == "alpha1") & (df_scenario1$T == 500), 7]
alpha1_upper_500 <- df_scenario1[ (df_scenario1$parameter == "alpha1") & (df_scenario1$T == 500), 11]
alpha1_lower_750 <- df_scenario1[ (df_scenario1$parameter == "alpha1") & (df_scenario1$T == 750), 7]
alpha1_upper_750 <- df_scenario1[ (df_scenario1$parameter == "alpha1") & (df_scenario1$T == 750), 11]
alpha1_lower_1000 <- df_scenario1[ (df_scenario1$parameter == "alpha1") & (df_scenario1$T == 1000), 7]
alpha1_upper_1000 <- df_scenario1[ (df_scenario1$parameter == "alpha1") & (df_scenario1$T == 1000), 11]
coverage_prob_alpha1_500 <- mean( (alpha1_lower_500 < true_alpha1) & (alpha1_upper_500 > true_alpha1) )
coverage_prob_alpha1_750 <- mean( (alpha1_lower_750 < true_alpha1) & (alpha1_upper_750 > true_alpha1) )
coverage_prob_alpha1_1000 <- mean( (alpha1_lower_1000 < true_alpha1) & (alpha1_upper_1000 > true_alpha1) )

beta1_lower_500 <- df_scenario1[ (df_scenario1$parameter == "beta1") & (df_scenario1$T == 500), 7]
beta1_upper_500 <- df_scenario1[ (df_scenario1$parameter == "beta1") & (df_scenario1$T == 500), 11]
beta1_lower_750 <- df_scenario1[ (df_scenario1$parameter == "beta1") & (df_scenario1$T == 750), 7]
beta1_upper_750 <- df_scenario1[ (df_scenario1$parameter == "beta1") & (df_scenario1$T == 750), 11]
beta1_lower_1000 <- df_scenario1[ (df_scenario1$parameter == "beta1") & (df_scenario1$T == 1000), 7]
beta1_upper_1000 <- df_scenario1[ (df_scenario1$parameter == "beta1") & (df_scenario1$T == 1000), 11]
coverage_prob_beta1_500 <- mean( (beta1_lower_500 < true_beta1) & (beta1_upper_500 > true_beta1) )
coverage_prob_beta1_750 <- mean( (beta1_lower_750 < true_beta1) & (beta1_upper_750 > true_beta1) )
coverage_prob_beta1_1000 <- mean( (beta1_lower_1000 < true_beta1) & (beta1_upper_1000 > true_beta1) )
```

```r
summary_scenario1 <- df_scenario1 %>% group_by(parameter, T) %>% summarise(
    true_value = first(true_value),
    bias = mean(bias, na.rm = TRUE),
    rmse = sqrt( mean(mse, na.rm = TRUE) ),
    X2.5. = mean(`X2.5.`, na.rm = TRUE),
    X25. = mean(`X25.`, na.rm = TRUE),
    X50. = mean(`X50.`, na.rm = TRUE),
    X75. = mean(`X75.`, na.rm = TRUE),
    X97.5. = mean(`X97.5.`, na.rm = TRUE),
    .groups = "drop"
)
summary_scenario1$coverage_prob <- c(
    coverage_prob_alpha0_500,
    coverage_prob_alpha0_750,
    coverage_prob_alpha0_1000,
    coverage_prob_alpha1_500,
    coverage_prob_alpha1_750,
    coverage_prob_alpha1_1000,
    coverage_prob_beta1_500,
    coverage_prob_beta1_750,
    coverage_prob_beta1_1000
)
summary_scenario1 <- summary_scenario1 %>% mutate(
    interval = paste0( "(", round(`X2.5.`, 2), ", ", round(`X97.5.`, 2), ")" )
)
summary_scenario1
```

```r
kbl(summary_scenario1[,c("parameter", "true_value", "T", "bias", "rmse", "interval", "coverage_prob")], format = "latex", booktabs = TRUE, digits = 4)
```
